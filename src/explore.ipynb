```python
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials


lz_uri = 'spotify:artist:4ZDoy7AWNgQVmX7T0u0B1j'

spotify = spotipy.Spotify(client_credentials_manager=SpotifyClientCredentials(client_id=client_id, client_secret=client_secret))
results = spotify.artist_top_tracks(lz_uri)

canciones = []
popularidad = []
duracion = []

for track in results['tracks'][:10]:
    
    canciones.append(track['name'])
    popularidad.append(track['popularity'])
    duracion.append(track['duration_ms'])

dic = {
    "canciones": canciones,
    "popularidad": popularidad,
    "duracion": duracion
}


data = pd.DataFrame(dic)

data
```

```python
# ¿Tiene relación la duración con la popularidad? ¿Podríamos decir que una canción que dure poco tiempo puede ser más popular que otra que dure más? Analízalo graficando un scatter plot y argumenta tu respuesta.

plt.scatter(data['duracion'], data['popularidad'])
plt.title("Duración vs Popularidad")
plt.xlabel("Duración (ms)")
plt.ylabel("Popularidad")
plt.show()
```

Se visualiza una relación directa entre duracion y popularidad, podemos ver que las canciones mas populares son las de menor duración y también las de menor duracion.
